```python
import os, json, re, random, torch
from PIL import Image
from tqdm import tqdm
from peft import PeftModel
from transformers import Qwen2_5_VLForConditionalGeneration, AutoProcessor, BitsAndBytesConfig
import matplotlib.pyplot as plt
import matplotlib.patches as patches

os.environ["TRANSFORMERS_NO_FLASH_ATTN"] = "1"

ADAPTER_DIR = "/workspace/insight_cxr_3b/adapter/"
MODEL_ID    = "Qwen/Qwen2.5-VL-3B-Instruct"

SYSTEM_PROMPT = (
    "You are an expert radiologist AI assistant. "
    "Given a chest X-ray image, provide accurate clinical findings. "
    "When referring to a specific region, output its bounding box "
    "immediately after using the format: <box>[x1, y1, x2, y2]</box> "
    "where coordinates are absolute pixels on a 448x448 image. "
    "Be factual. Do not hallucinate findings not visible in the image."
)

# ── Load 3B model ─────────────────────────────────────────────
print("Loading 3B model...")
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_compute_dtype=torch.bfloat16,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_use_double_quant=True,
)
base_model = Qwen2_5_VLForConditionalGeneration.from_pretrained(
    MODEL_ID,
    quantization_config=bnb_config,
    device_map="auto",
    torch_dtype=torch.bfloat16,
    attn_implementation="eager",
)
ft_model  = PeftModel.from_pretrained(base_model, ADAPTER_DIR)
ft_model.eval()
processor = AutoProcessor.from_pretrained(ADAPTER_DIR)
print("3B model ready.")

# ── Inference function ────────────────────────────────────────
def run_inference(image_path, question):
    try:
        img = Image.open(image_path).convert("RGB")
    except:
        return ""
    messages = [
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user", "content": [
            {"type": "image"},
            {"type": "text", "text": question},
        ]},
    ]
    prompt = processor.apply_chat_template(
        messages, tokenize=False, add_generation_prompt=True
    )
    inputs = processor(
        text=[prompt], images=[img],
        return_tensors="pt", padding=True
    ).to("cuda")
    with torch.no_grad():
        out = ft_model.generate(
            **inputs, max_new_tokens=300,
            do_sample=False, repetition_penalty=1.1,
        )
    return processor.tokenizer.decode(
        out[0][inputs["input_ids"].shape[1]:],
        skip_special_tokens=True,
    )

def compute_iou(b1, b2):
    x1 = max(b1[0], b2[0]); y1 = max(b1[1], b2[1])
    x2 = min(b1[2], b2[2]); y2 = min(b1[3], b2[3])
    inter = max(0, x2-x1) * max(0, y2-y1)
    a1 = (b1[2]-b1[0]) * (b1[3]-b1[1])
    a2 = (b2[2]-b2[0]) * (b2[3]-b2[1])
    union = a1 + a2 - inter
    return inter / union if union > 0 else 0.0

# ── Test examples visualization ───────────────────────────────
random.seed(42)
with open("/workspace/insight_cxr/processed/test.jsonl") as f:
    all_test = [json.loads(l) for l in f if l.strip()]
samples = random.sample(all_test, 5)

fig, axes = plt.subplots(5, 2, figsize=(14, 35))

for i, sample in enumerate(samples):
    response   = run_inference(sample["image_path"], sample["question"])
    clean      = re.sub(r'<think>.*?</think>', '', response, flags=re.DOTALL).strip()
    answer     = re.sub(r'<box>\[.*?\]</box>', '', clean).strip()
    pred_boxes = re.findall(r'<box>\[(\d+),\s*(\d+),\s*(\d+),\s*(\d+)\]</box>', response)

    iou = 0.0
    if pred_boxes and sample["bboxes"]:
        pred = [int(x) for x in pred_boxes[0]]
        iou  = compute_iou(pred, sample["bboxes"][0])

    print(f"\n[Sample {i+1}]")
    print(f"Q       : {sample['question']}")
    print(f"GT      : {sample['answer']}")
    print(f"Pred    : {answer}")
    print(f"GT bbox : {sample['bboxes']}")
    print(f"Pred box: {pred_boxes}")
    print(f"IoU     : {iou:.3f}")

    img = Image.open(sample["image_path"]).convert("RGB")

    axes[i][0].imshow(img)
    axes[i][0].set_title(f"GT | {sample['answer'][:40]}", fontsize=8)
    axes[i][0].axis("off")
    for bbox in sample["bboxes"]:
        x1,y1,x2,y2 = bbox
        axes[i][0].add_patch(patches.Rectangle(
            (x1,y1), x2-x1, y2-y1,
            linewidth=2, edgecolor="#4BFF91", facecolor="none"
        ))

    axes[i][1].imshow(img)
    axes[i][1].set_title(f"3B Pred IoU={iou:.3f} | {answer[:40]}", fontsize=8)
    axes[i][1].axis("off")
    for bbox in pred_boxes:
        x1,y1,x2,y2 = [int(x) for x in bbox]
        axes[i][1].add_patch(patches.Rectangle(
            (x1,y1), x2-x1, y2-y1,
            linewidth=2, edgecolor="#FF4B4B", facecolor="none"
        ))

plt.tight_layout()
plt.savefig("/workspace/insight_cxr_3b/test_examples_3b.png", dpi=150, bbox_inches="tight")
plt.show()
print("Saved → /workspace/insight_cxr_3b/test_examples_3b.png")

# ── Full evaluation 500 samples ───────────────────────────────
test_samples = random.sample(all_test, 500)
print(f"\nEvaluating on 500 test samples...")

predictions = []
references  = []
iou_scores  = []
bbox_found  = 0

for i, sample in enumerate(tqdm(test_samples)):
    response = run_inference(sample["image_path"], sample["question"])
    if not response:
        continue

    clean      = re.sub(r'<think>.*?</think>', '', response, flags=re.DOTALL).strip()
    answer     = re.sub(r'<box>\[.*?\]</box>', '', clean).strip()
    pred_boxes = re.findall(r'<box>\[(\d+),\s*(\d+),\s*(\d+),\s*(\d+)\]</box>', response)

    predictions.append(answer)
    references.append(sample["answer"])

    if pred_boxes and sample["bboxes"]:
        bbox_found += 1
        pred = [int(x) for x in pred_boxes[0]]
        iou_scores.append(compute_iou(pred, sample["bboxes"][0]))
    elif sample["bboxes"]:
        iou_scores.append(0.0)

    if (i+1) % 100 == 0:
        print(f"[{i+1}/500] mean IoU: {sum(iou_scores)/len(iou_scores):.3f}  bbox_found: {bbox_found}/{i+1}")

# Metrics
from nltk.translate.bleu_score import corpus_bleu, SmoothingFunction
from rouge_score import rouge_scorer

smooth    = SmoothingFunction().method1
tok_preds = [p.lower().split() for p in predictions]
tok_refs  = [[r.lower().split()] for r in references]
bleu1 = corpus_bleu(tok_refs, tok_preds, weights=(1,0,0,0), smoothing_function=smooth)
bleu4 = corpus_bleu(tok_refs, tok_preds, weights=(.25,.25,.25,.25), smoothing_function=smooth)

scorer  = rouge_scorer.RougeScorer(["rougeL"], use_stemmer=True)
rouge_l = sum(scorer.score(r, p)["rougeL"].fmeasure
              for r, p in zip(references, predictions)) / len(predictions)
mean_iou = sum(iou_scores) / len(iou_scores) if iou_scores else 0.0

print(f"\n{'='*55}")
print(f"FINAL EVALUATION — Qwen2.5-VL-3B  (n=500 test samples)")
print(f"{'='*55}")
print(f"BLEU-1     : {bleu1:.4f}")
print(f"BLEU-4     : {bleu4:.4f}")
print(f"ROUGE-L    : {rouge_l:.4f}")
print(f"Mean IoU   : {mean_iou:.4f}")
print(f"Bbox found : {bbox_found}/{len(iou_scores)} ({100*bbox_found/len(iou_scores):.1f}%)")
print(f"{'='*55}")

print(f"\n{'='*55}")
print(f"2B vs 3B COMPARISON")
print(f"{'='*55}")
print(f"Metric        2B Model    3B Model")
print(f"BLEU-1        0.0302      {bleu1:.4f}")
print(f"ROUGE-L       0.4138      {rouge_l:.4f}")
print(f"Mean IoU      0.2299      {mean_iou:.4f}")
print(f"Val Loss      3.839       3.800")
print(f"{'='*55}")
```

```python
import json, re, torch
from PIL import Image
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import os

ADAPTER_DIR = "/workspace/insight_cxr_3b/adapter/"
SYSTEM_PROMPT = (
    "You are an expert radiologist AI assistant. "
    "Given a chest X-ray image, provide accurate clinical findings. "
    "When referring to a specific region, output its bounding box "
    "immediately after using the format: <box>[x1, y1, x2, y2]</box> "
    "where coordinates are absolute pixels on a 448x448 image. "
    "Be factual. Do not hallucinate findings not visible in the image."
)

def run_inference(image_path, question):
    img = Image.open(image_path).convert("RGB")
    messages = [
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user", "content": [
            {"type": "image"},
            {"type": "text", "text": question},
        ]},
    ]
    prompt = processor.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
    inputs = processor(text=[prompt], images=[img], return_tensors="pt", padding=True).to("cuda")
    with torch.no_grad():
        out = ft_model.generate(**inputs, max_new_tokens=300, do_sample=False, repetition_penalty=1.1)
    return processor.tokenizer.decode(out[0][inputs["input_ids"].shape[1]:], skip_special_tokens=True)

# Pick 4 samples
chosen_indices = [0, 5, 7, 12, 30]
samples = [all_test[i] for i in chosen_indices]

os.makedirs("/workspace/prototype_screenshots", exist_ok=True)

for idx, sample in enumerate(samples):
    response = run_inference(sample["image_path"], sample["question"])

    # Parse
    think = re.search(r'<think>(.*?)</think>', response, re.DOTALL)
    reasoning = think.group(1).strip()[:300] if think else ""
    clean = re.sub(r'<think>.*?</think>', '', response, flags=re.DOTALL).strip()
    answer = re.sub(r'<box>\[.*?\]</box>', '', clean).strip()
    pred_boxes = re.findall(r'<box>\[(\d+),\s*(\d+),\s*(\d+),\s*(\d+)\]</box>', response)

    img = Image.open(sample["image_path"]).convert("RGB")

    # ── Figure layout ──
    fig = plt.figure(figsize=(14, 6))
    fig.patch.set_facecolor('#0D1B2A')

    # Left — image with boxes
    ax1 = fig.add_axes([0.02, 0.05, 0.42, 0.88])
    ax1.imshow(img, cmap='gray')
    ax1.set_facecolor('#0D1B2A')
    ax1.axis('off')

    # Ground truth bbox — green
    for bbox in sample["bboxes"]:
        x1,y1,x2,y2 = bbox
        ax1.add_patch(patches.Rectangle((x1,y1), x2-x1, y2-y1,
            linewidth=2.5, edgecolor='#38E54D', facecolor='none'))
        ax1.text(x1, max(y1-5,0), 'GT', color='#38E54D', fontsize=9, fontweight='bold',
            bbox=dict(boxstyle='round,pad=0.2', facecolor='#0D1B2A', alpha=0.8))

    # Predicted bbox — red
    for bbox in pred_boxes:
        x1,y1,x2,y2 = [int(x) for x in bbox]
        ax1.add_patch(patches.Rectangle((x1,y1), x2-x1, y2-y1,
            linewidth=2.5, edgecolor='#FF4B4B', facecolor='none'))
        ax1.text(x2+2, y1, 'Pred', color='#FF4B4B', fontsize=9, fontweight='bold',
            bbox=dict(boxstyle='round,pad=0.2', facecolor='#0D1B2A', alpha=0.8))

    ax1.set_title('🟢 GT   🔴 Predicted', color='white', fontsize=10, pad=6)

    # Right — text panel
    ax2 = fig.add_axes([0.46, 0.05, 0.52, 0.88])
    ax2.set_facecolor('#132338')
    ax2.axis('off')

    y_pos = 0.97
    def add_text(ax, text, y, color, size=9.5, bold=False, wrap=60):
        weight = 'bold' if bold else 'normal'
        lines = [text[i:i+wrap] for i in range(0, len(text), wrap)]
        for line in lines:
            ax.text(0.02, y, line, transform=ax.transAxes,
                color=color, fontsize=size, fontweight=weight,
                verticalalignment='top', fontfamily='monospace')
            y -= 0.07
        return y

    y_pos = add_text(ax2, 'QUESTION', y_pos, '#00C2FF', size=10, bold=True)
    y_pos = add_text(ax2, sample['question'][:100], y_pos, '#A8B8CC', size=9, wrap=55)
    y_pos -= 0.04

    y_pos = add_text(ax2, 'REASONING (Chain-of-Thought)', y_pos, '#B48FFF', size=10, bold=True)
    y_pos = add_text(ax2, reasoning[:250] + ('...' if len(reasoning)>250 else ''), y_pos, '#778899', size=8, wrap=65)
    y_pos -= 0.04

    y_pos = add_text(ax2, 'PREDICTED ANSWER', y_pos, '#38E54D', size=10, bold=True)
    y_pos = add_text(ax2, answer[:100], y_pos, '#CCDDCC', size=9, wrap=55)
    y_pos -= 0.04

    y_pos = add_text(ax2, 'GROUND TRUTH ANSWER', y_pos, '#FFD166', size=10, bold=True)
    y_pos = add_text(ax2, str(sample['answer'])[:100], y_pos, '#CCCC99', size=9, wrap=55)
    y_pos -= 0.04

    if pred_boxes:
        y_pos = add_text(ax2, 'PREDICTED BBOX', y_pos, '#FF8C42', size=10, bold=True)
        y_pos = add_text(ax2, f'<box>{list(pred_boxes[0])}</box>', y_pos, '#FF8C42', size=9)
        y_pos = add_text(ax2, f'GT bbox: {sample["bboxes"][0]}', y_pos, '#FFD166', size=9)

    plt.suptitle(f'INSIGHT-CXR  |  Working Prototype  |  Sample {idx+1}',
        color='white', fontsize=12, fontweight='bold', y=0.98,
        bbox=dict(facecolor='#0A2540', alpha=0.8, pad=4))

    path = f"/workspace/prototype_screenshots/sample_{idx+1}.png"
    plt.savefig(path, dpi=150, bbox_inches='tight', facecolor='#0D1B2A')
    plt.show()
    print(f"Saved → {path}")
    plt.close()

print("\nAll screenshots saved to /workspace/prototype_screenshots/")
```

```python
import json, re, torch
from PIL import Image
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import os

os.makedirs("/workspace/prototype_screenshots", exist_ok=True)

chosen_indices = [0, 1, 2, 12, 13]
samples = [all_test[i] for i in chosen_indices]

for idx, sample in enumerate(samples):
    response = run_inference(sample["image_path"], sample["question"])

    think = re.search(r'<think>(.*?)</think>', response, re.DOTALL)
    reasoning = think.group(1).strip()[:280] if think else ""
    clean = re.sub(r'<think>.*?</think>', '', response, flags=re.DOTALL).strip()
    answer = re.sub(r'<box>\[.*?\]</box>', '', clean).strip()
    pred_boxes = re.findall(r'<box>\[(\d+),\s*(\d+),\s*(\d+),\s*(\d+)\]</box>', response)

    img = Image.open(sample["image_path"]).convert("RGB")

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5.5))
    fig.patch.set_facecolor('white')

    # ── LEFT: X-ray with boxes ──
    ax1.imshow(img, cmap='gray')
    ax1.axis('off')
    ax1.set_facecolor('black')

    for bbox in sample["bboxes"]:
        x1,y1,x2,y2 = bbox
        ax1.add_patch(patches.Rectangle((x1,y1), x2-x1, y2-y1,
            linewidth=2, edgecolor='#2ecc71', facecolor='none'))
        ax1.text(x1, max(y1-6,0), 'Ground Truth',
            color='#2ecc71', fontsize=8, fontweight='bold',
            bbox=dict(boxstyle='round,pad=0.2', facecolor='black', alpha=0.6))

    for bbox in pred_boxes:
        x1,y1,x2,y2 = [int(x) for x in bbox]
        ax1.add_patch(patches.Rectangle((x1,y1), x2-x1, y2-y1,
            linewidth=2, edgecolor='#e74c3c', facecolor='none', linestyle='--'))
        ax1.text(x1, y2+8, 'Predicted',
            color='#e74c3c', fontsize=8, fontweight='bold',
            bbox=dict(boxstyle='round,pad=0.2', facecolor='black', alpha=0.6))

    ax1.set_title(f'Sample {idx+1}  |  ── Ground Truth   - - Predicted',
        fontsize=10, color='#333333', pad=8)

    # ── RIGHT: Text panel ──
    ax2.set_facecolor('white')
    ax2.axis('off')

    content = []

    content.append(('QUESTION', '#2c3e50', 10, True))
    content.append((sample['question'][:120], '#555555', 9, False))
    content.append(('', '', 9, False))

    content.append(('CLINICAL REASONING', '#8e44ad', 10, True))
    short_reason = reasoning[:220] + ('...' if len(reasoning) > 220 else '')
    content.append((short_reason, '#666666', 8.5, False))
    content.append(('', '', 9, False))

    content.append(('PREDICTED ANSWER', '#27ae60', 10, True))
    content.append((answer[:120], '#333333', 9, False))
    content.append(('', '', 9, False))

    content.append(('GROUND TRUTH ANSWER', '#e67e22', 10, True))
    content.append((str(sample['answer'])[:120], '#333333', 9, False))

    if pred_boxes:
        content.append(('', '', 9, False))
        content.append(('BOUNDING BOX', '#c0392b', 10, True))
        pb = [int(x) for x in pred_boxes[0]]
        gt = sample['bboxes'][0]
        content.append((f'Predicted : [{pb[0]}, {pb[1]}, {pb[2]}, {pb[3]}]', '#c0392b', 9, False))
        content.append((f'Ground Truth: {gt}', '#e67e22', 9, False))

    y = 0.97
    for text, color, size, bold in content:
        if text == '':
            y -= 0.04
            continue
        weight = 'bold' if bold else 'normal'
        # Word wrap
        words = text.split()
        line = ''
        for word in words:
            if len(line + word) > 58:
                ax2.text(0.02, y, line, transform=ax2.transAxes,
                    color=color, fontsize=size, fontweight=weight,
                    verticalalignment='top', fontfamily='DejaVu Sans')
                y -= 0.075
                line = word + ' '
            else:
                line += word + ' '
        if line:
            ax2.text(0.02, y, line.strip(), transform=ax2.transAxes,
                color=color, fontsize=size, fontweight=weight,
                verticalalignment='top', fontfamily='DejaVu Sans')
            y -= 0.075

    # Thin divider line between panels
    fig.add_artist(plt.Line2D([0.47, 0.47], [0.05, 0.95],
        transform=fig.transFigure, color='#dddddd', linewidth=1))

    plt.suptitle('INSIGHT-CXR  —  Working Prototype',
        fontsize=13, fontweight='bold', color='#2c3e50', y=1.01)

    plt.tight_layout(pad=1.5)
    path = f"/workspace/prototype_screenshots/sample_{idx+1}.png"
    plt.savefig(path, dpi=150, bbox_inches='tight',
        facecolor='white', edgecolor='none')
    plt.show()
    print(f"Saved → {path}")
    plt.close()

print("\nDone. Download from /workspace/prototype_screenshots/")
```
