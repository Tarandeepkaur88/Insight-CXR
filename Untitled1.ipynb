```python
import os, json, re, random, torch
from PIL import Image
from tqdm import tqdm
from peft import PeftModel
from transformers import Qwen2_5_VLForConditionalGeneration, AutoProcessor, BitsAndBytesConfig
import matplotlib.pyplot as plt
import matplotlib.patches as patches

os.environ["TRANSFORMERS_NO_FLASH_ATTN"] = "1"

ADAPTER_DIR = "/workspace/insight_cxr_3b/adapter/"
MODEL_ID    = "Qwen/Qwen2.5-VL-3B-Instruct"

SYSTEM_PROMPT = (
    "You are an expert radiologist AI assistant. "
    "Given a chest X-ray image, provide accurate clinical findings. "
    "When referring to a specific region, output its bounding box "
    "immediately after using the format: <box>[x1, y1, x2, y2]</box> "
    "where coordinates are absolute pixels on a 448x448 image. "
    "Be factual. Do not hallucinate findings not visible in the image."
)

# ── Load 3B model ─────────────────────────────────────────────
print("Loading 3B model...")
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_compute_dtype=torch.bfloat16,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_use_double_quant=True,
)
base_model = Qwen2_5_VLForConditionalGeneration.from_pretrained(
    MODEL_ID,
    quantization_config=bnb_config,
    device_map="auto",
    torch_dtype=torch.bfloat16,
    attn_implementation="eager",
)
ft_model  = PeftModel.from_pretrained(base_model, ADAPTER_DIR)
ft_model.eval()
processor = AutoProcessor.from_pretrained(ADAPTER_DIR)
print("3B model ready.")

# ── Inference function ────────────────────────────────────────
def run_inference(image_path, question):
    try:
        img = Image.open(image_path).convert("RGB")
    except:
        return ""
    messages = [
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user", "content": [
            {"type": "image"},
            {"type": "text", "text": question},
        ]},
    ]
    prompt = processor.apply_chat_template(
        messages, tokenize=False, add_generation_prompt=True
    )
    inputs = processor(
        text=[prompt], images=[img],
        return_tensors="pt", padding=True
    ).to("cuda")
    with torch.no_grad():
        out = ft_model.generate(
            **inputs, max_new_tokens=300,
            do_sample=False, repetition_penalty=1.1,
        )
    return processor.tokenizer.decode(
        out[0][inputs["input_ids"].shape[1]:],
        skip_special_tokens=True,
    )

def compute_iou(b1, b2):
    x1 = max(b1[0], b2[0]); y1 = max(b1[1], b2[1])
    x2 = min(b1[2], b2[2]); y2 = min(b1[3], b2[3])
    inter = max(0, x2-x1) * max(0, y2-y1)
    a1 = (b1[2]-b1[0]) * (b1[3]-b1[1])
    a2 = (b2[2]-b2[0]) * (b2[3]-b2[1])
    union = a1 + a2 - inter
    return inter / union if union > 0 else 0.0

# ── Test examples visualization ───────────────────────────────
random.seed(42)
with open("/workspace/insight_cxr/processed/test.jsonl") as f:
    all_test = [json.loads(l) for l in f if l.strip()]
samples = random.sample(all_test, 5)

fig, axes = plt.subplots(5, 2, figsize=(14, 35))

for i, sample in enumerate(samples):
    response   = run_inference(sample["image_path"], sample["question"])
    clean      = re.sub(r'<think>.*?</think>', '', response, flags=re.DOTALL).strip()
    answer     = re.sub(r'<box>\[.*?\]</box>', '', clean).strip()
    pred_boxes = re.findall(r'<box>\[(\d+),\s*(\d+),\s*(\d+),\s*(\d+)\]</box>', response)

    iou = 0.0
    if pred_boxes and sample["bboxes"]:
        pred = [int(x) for x in pred_boxes[0]]
        iou  = compute_iou(pred, sample["bboxes"][0])

    print(f"\n[Sample {i+1}]")
    print(f"Q       : {sample['question']}")
    print(f"GT      : {sample['answer']}")
    print(f"Pred    : {answer}")
    print(f"GT bbox : {sample['bboxes']}")
    print(f"Pred box: {pred_boxes}")
    print(f"IoU     : {iou:.3f}")

    img = Image.open(sample["image_path"]).convert("RGB")

    axes[i][0].imshow(img)
    axes[i][0].set_title(f"GT | {sample['answer'][:40]}", fontsize=8)
    axes[i][0].axis("off")
    for bbox in sample["bboxes"]:
        x1,y1,x2,y2 = bbox
        axes[i][0].add_patch(patches.Rectangle(
            (x1,y1), x2-x1, y2-y1,
            linewidth=2, edgecolor="#4BFF91", facecolor="none"
        ))

    axes[i][1].imshow(img)
    axes[i][1].set_title(f"3B Pred IoU={iou:.3f} | {answer[:40]}", fontsize=8)
    axes[i][1].axis("off")
    for bbox in pred_boxes:
        x1,y1,x2,y2 = [int(x) for x in bbox]
        axes[i][1].add_patch(patches.Rectangle(
            (x1,y1), x2-x1, y2-y1,
            linewidth=2, edgecolor="#FF4B4B", facecolor="none"
        ))

plt.tight_layout()
plt.savefig("/workspace/insight_cxr_3b/test_examples_3b.png", dpi=150, bbox_inches="tight")
plt.show()
print("Saved → /workspace/insight_cxr_3b/test_examples_3b.png")

# ── Full evaluation 500 samples ───────────────────────────────
test_samples = random.sample(all_test, 500)
print(f"\nEvaluating on 500 test samples...")

predictions = []
references  = []
iou_scores  = []
bbox_found  = 0

for i, sample in enumerate(tqdm(test_samples)):
    response = run_inference(sample["image_path"], sample["question"])
    if not response:
        continue

    clean      = re.sub(r'<think>.*?</think>', '', response, flags=re.DOTALL).strip()
    answer     = re.sub(r'<box>\[.*?\]</box>', '', clean).strip()
    pred_boxes = re.findall(r'<box>\[(\d+),\s*(\d+),\s*(\d+),\s*(\d+)\]</box>', response)

    predictions.append(answer)
    references.append(sample["answer"])

    if pred_boxes and sample["bboxes"]:
        bbox_found += 1
        pred = [int(x) for x in pred_boxes[0]]
        iou_scores.append(compute_iou(pred, sample["bboxes"][0]))
    elif sample["bboxes"]:
        iou_scores.append(0.0)

    if (i+1) % 100 == 0:
        print(f"[{i+1}/500] mean IoU: {sum(iou_scores)/len(iou_scores):.3f}  bbox_found: {bbox_found}/{i+1}")

# Metrics
from nltk.translate.bleu_score import corpus_bleu, SmoothingFunction
from rouge_score import rouge_scorer

smooth    = SmoothingFunction().method1
tok_preds = [p.lower().split() for p in predictions]
tok_refs  = [[r.lower().split()] for r in references]
bleu1 = corpus_bleu(tok_refs, tok_preds, weights=(1,0,0,0), smoothing_function=smooth)
bleu4 = corpus_bleu(tok_refs, tok_preds, weights=(.25,.25,.25,.25), smoothing_function=smooth)

scorer  = rouge_scorer.RougeScorer(["rougeL"], use_stemmer=True)
rouge_l = sum(scorer.score(r, p)["rougeL"].fmeasure
              for r, p in zip(references, predictions)) / len(predictions)
mean_iou = sum(iou_scores) / len(iou_scores) if iou_scores else 0.0

print(f"\n{'='*55}")
print(f"FINAL EVALUATION — Qwen2.5-VL-3B  (n=500 test samples)")
print(f"{'='*55}")
print(f"BLEU-1     : {bleu1:.4f}")
print(f"BLEU-4     : {bleu4:.4f}")
print(f"ROUGE-L    : {rouge_l:.4f}")
print(f"Mean IoU   : {mean_iou:.4f}")
print(f"Bbox found : {bbox_found}/{len(iou_scores)} ({100*bbox_found/len(iou_scores):.1f}%)")
print(f"{'='*55}")

print(f"\n{'='*55}")
print(f"2B vs 3B COMPARISON")
print(f"{'='*55}")
print(f"Metric        2B Model    3B Model")
print(f"BLEU-1        0.0302      {bleu1:.4f}")
print(f"ROUGE-L       0.4138      {rouge_l:.4f}")
print(f"Mean IoU      0.2299      {mean_iou:.4f}")
print(f"Val Loss      3.839       3.800")
print(f"{'='*55}")
```

```python
# ============================================================
# ONE-CELL COMPLETE EVALUATION
# Qwen2.5-VL-3B + LoRA
#
# Metrics:
#   BLEU-1
#   BLEU-4
#   ROUGE-L
#   BERTScore
#   Mean IoU
#   Median IoU
#   IoU@0.50
#   IoU@0.75
#   IoU@0.90
#   BBox Found %
#   Bootstrap 95% CI
#
# Saves:
#   per-image predictions CSV
#   final metrics JSON
#   summary CSV
# ============================================================

# ------------------------------------------------------------
# 0. INSTALL BASIC PACKAGES
# ------------------------------------------------------------

import sys
import subprocess
import importlib.util

packages = {
    "rouge_score": "rouge-score",
    "bert_score": "bert-score",
    "nltk": "nltk",
    "pandas": "pandas",
    "scipy": "scipy",
    "tqdm": "tqdm",
}

for module, package in packages.items():
    if importlib.util.find_spec(module) is None:
        print(f"Installing {package} ...")
        subprocess.check_call(
            [sys.executable, "-m", "pip", "install", "-q", package]
        )

# ------------------------------------------------------------
# 1. IMPORTS
# ------------------------------------------------------------

import os
import re
import json
import random
import warnings

import numpy as np
import pandas as pd
import torch

from PIL import Image
from tqdm.auto import tqdm

from torch.utils.data import Dataset

from transformers import (
    Qwen2_5_VLForConditionalGeneration,
    AutoProcessor,
    BitsAndBytesConfig,
)

from peft import PeftModel

from nltk.translate.bleu_score import (
    sentence_bleu,
    SmoothingFunction,
)

from rouge_score import rouge_scorer

warnings.filterwarnings("ignore")


# ------------------------------------------------------------
# 2. CONFIGURATION
# ------------------------------------------------------------

MODEL_ID = "Qwen/Qwen2.5-VL-3B-Instruct"

ADAPTER = "/workspace/insight_cxr_3b/adapter/"

DATA_DIR = "/workspace/insight_cxr/processed/"

TEST_JSONL = os.path.join(
    DATA_DIR,
    "test.jsonl"
)

OUTPUT_DIR = "/workspace/insight_cxr_3b/evaluation/"

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)

MAX_TEST_SAMPLES = 500

MAX_NEW_TOKENS = 512

IMAGE_SIZE = 448

SEED = 42

RUN_BERTSCORE = True


# ------------------------------------------------------------
# 3. SEED
# ------------------------------------------------------------

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)


print("=" * 75)
print("QWEN2.5-VL-3B CXR EVALUATION")
print("=" * 75)

print("CUDA:", torch.cuda.is_available())

if torch.cuda.is_available():
    print(
        "GPU:",
        torch.cuda.get_device_name(0)
    )

print()
print("Test file:", TEST_JSONL)
print("Adapter:", ADAPTER)

if not os.path.exists(TEST_JSONL):
    raise FileNotFoundError(
        f"Test file not found:\n{TEST_JSONL}"
    )

if not os.path.exists(ADAPTER):
    raise FileNotFoundError(
        f"LoRA adapter not found:\n{ADAPTER}"
    )


# ------------------------------------------------------------
# 4. LOAD TEST DATA
# ------------------------------------------------------------

class CXRTestDataset(Dataset):

    def __init__(self, jsonl_path):

        self.samples = []

        with open(
            jsonl_path,
            "r"
        ) as f:

            for line in f:

                if line.strip():

                    self.samples.append(
                        json.loads(line)
                    )

        print(
            f"\nLoaded {len(self.samples):,} samples"
        )

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):

        s = self.samples[idx]

        try:

            image = Image.open(
                s["image_path"]
            ).convert("RGB")

        except Exception:

            print(
                f"WARNING: image failed: "
                f"{s['image_path']}"
            )

            image = Image.new(
                "RGB",
                (IMAGE_SIZE, IMAGE_SIZE),
                128
            )

        return {
            "image": image,
            "image_path": s["image_path"],
            "question": s["question"],
            "answer": s["answer"],
            "bboxes": s.get("bboxes", [])
        }


dataset = CXRTestDataset(
    TEST_JSONL
)

N = min(
    len(dataset),
    MAX_TEST_SAMPLES
)

print(
    f"Evaluating {N} samples"
)


# ------------------------------------------------------------
# 5. LOAD MODEL
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("LOADING MODEL")
print("=" * 75)

bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_compute_dtype=torch.bfloat16,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_use_double_quant=True,
)

base_model = (
    Qwen2_5_VLForConditionalGeneration
    .from_pretrained(
        MODEL_ID,
        quantization_config=bnb_config,
        device_map="auto",
        torch_dtype=torch.bfloat16,
        attn_implementation="eager",
    )
)

base_model.config.use_cache = True

model = PeftModel.from_pretrained(
    base_model,
    ADAPTER
)

model.eval()

processor = AutoProcessor.from_pretrained(
    ADAPTER,
    min_pixels=256 * 28 * 28,
    max_pixels=1280 * 28 * 28,
    padding_side="left",
)

if processor.tokenizer.pad_token is None:

    processor.tokenizer.pad_token = (
        processor.tokenizer.eos_token
    )

print("Model + LoRA loaded successfully.")


# ------------------------------------------------------------
# 6. SAME SYSTEM PROMPT USED DURING TRAINING
# ------------------------------------------------------------

SYSTEM_PROMPT = (
    "You are an expert radiologist AI assistant. "
    "Given a chest X-ray image, provide accurate clinical findings. "
    "When referring to a specific region, output its bounding box "
    "immediately after using the format: <box>[x1, y1, x2, y2]</box> "
    "where coordinates are absolute pixels on a 448x448 image. "
    "Be factual. Do not hallucinate findings not visible in the image."
)


def build_prompt(question):

    return [
        {
            "role": "system",
            "content": SYSTEM_PROMPT
        },
        {
            "role": "user",
            "content": [
                {
                    "type": "image"
                },
                {
                    "type": "text",
                    "text": question
                }
            ]
        }
    ]


# ------------------------------------------------------------
# 7. BBOX FUNCTIONS
# ------------------------------------------------------------

BBOX_PATTERN = re.compile(
    r"<box>\s*\[\s*"
    r"(-?\d+(?:\.\d+)?)\s*,\s*"
    r"(-?\d+(?:\.\d+)?)\s*,\s*"
    r"(-?\d+(?:\.\d+)?)\s*,\s*"
    r"(-?\d+(?:\.\d+)?)"
    r"\s*\]\s*</box>",
    re.IGNORECASE
)


def parse_bboxes(text):

    boxes = []

    for match in BBOX_PATTERN.findall(text):

        boxes.append([
            float(match[0]),
            float(match[1]),
            float(match[2]),
            float(match[3]),
        ])

    return boxes


def normalize_box(box):

    if box is None:
        return None

    x1, y1, x2, y2 = box

    x1 = max(
        0,
        min(IMAGE_SIZE, x1)
    )

    y1 = max(
        0,
        min(IMAGE_SIZE, y1)
    )

    x2 = max(
        0,
        min(IMAGE_SIZE, x2)
    )

    y2 = max(
        0,
        min(IMAGE_SIZE, y2)
    )

    if x2 < x1:
        x1, x2 = x2, x1

    if y2 < y1:
        y1, y2 = y2, y1

    return [
        x1,
        y1,
        x2,
        y2
    ]


def calculate_iou(box1, box2):

    if box1 is None or box2 is None:
        return 0.0

    x_left = max(
        box1[0],
        box2[0]
    )

    y_top = max(
        box1[1],
        box2[1]
    )

    x_right = min(
        box1[2],
        box2[2]
    )

    y_bottom = min(
        box1[3],
        box2[3]
    )

    inter_w = max(
        0,
        x_right - x_left
    )

    inter_h = max(
        0,
        y_bottom - y_top
    )

    intersection = (
        inter_w * inter_h
    )

    area1 = (
        max(
            0,
            box1[2] - box1[0]
        )
        *
        max(
            0,
            box1[3] - box1[1]
        )
    )

    area2 = (
        max(
            0,
            box2[2] - box2[0]
        )
        *
        max(
            0,
            box2[3] - box2[1]
        )
    )

    union = (
        area1 +
        area2 -
        intersection
    )

    if union <= 0:
        return 0.0

    return intersection / union


def best_box_iou(
    predicted_boxes,
    ground_truth_boxes
):

    if (
        len(predicted_boxes) == 0
        or
        len(ground_truth_boxes) == 0
    ):
        return 0.0

    best = 0.0

    for pred in predicted_boxes:

        pred = normalize_box(pred)

        for gt in ground_truth_boxes:

            gt = normalize_box(gt)

            score = calculate_iou(
                pred,
                gt
            )

            best = max(
                best,
                score
            )

    return best


# ------------------------------------------------------------
# 8. CLEAN REPORT
# ------------------------------------------------------------

def clean_report(text):

    text = re.sub(
        r"<think>.*?</think>",
        " ",
        text,
        flags=re.IGNORECASE | re.DOTALL
    )

    text = re.sub(
        r"<box>.*?</box>",
        " ",
        text,
        flags=re.IGNORECASE | re.DOTALL
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    )

    return text.strip()


# ------------------------------------------------------------
# 9. GENERATION
# ------------------------------------------------------------

@torch.no_grad()
def generate_prediction(
    image,
    question
):

    messages = build_prompt(
        question
    )

    prompt = processor.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = processor(
        text=[prompt],
        images=[image],
        return_tensors="pt",
        padding=True
    )

    inputs = {
        k: (
            v.to(model.device)
            if hasattr(v, "to")
            else v
        )
        for k, v in inputs.items()
    }

    generated_ids = model.generate(
        **inputs,
        max_new_tokens=MAX_NEW_TOKENS,
        do_sample=False,
        num_beams=1,
    )

    input_len = (
        inputs["input_ids"].shape[1]
    )

    generated_tokens = generated_ids[
        0,
        input_len:
    ]

    output = processor.tokenizer.decode(
        generated_tokens,
        skip_special_tokens=True
    )

    return output.strip()


# ------------------------------------------------------------
# 10. TEXT METRICS
# ------------------------------------------------------------

smooth = SmoothingFunction().method1

rouge_metric = rouge_scorer.RougeScorer(
    ["rougeL"],
    use_stemmer=True
)


def calculate_bleu(
    reference,
    prediction
):

    ref = reference.lower().split()
    pred = prediction.lower().split()

    if len(pred) == 0:
        return 0.0, 0.0

    bleu1 = sentence_bleu(
        [ref],
        pred,
        weights=(1, 0, 0, 0),
        smoothing_function=smooth
    )

    bleu4 = sentence_bleu(
        [ref],
        pred,
        weights=(
            0.25,
            0.25,
            0.25,
            0.25
        ),
        smoothing_function=smooth
    )

    return bleu1, bleu4


def calculate_rouge(
    reference,
    prediction
):

    score = rouge_metric.score(
        reference,
        prediction
    )

    return score[
        "rougeL"
    ].fmeasure


# ------------------------------------------------------------
# 11. EVALUATION LOOP
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("GENERATING TEST PREDICTIONS")
print("=" * 75)

records = []

for idx in tqdm(
    range(N),
    desc="Test evaluation"
):

    sample = dataset[idx]

    reference = str(
        sample["answer"]
    )

    gt_boxes = [
        normalize_box(box)
        for box in sample["bboxes"]
    ]

    # Generate
    raw_prediction = generate_prediction(
        sample["image"],
        sample["question"]
    )

    # Parse boxes
    predicted_boxes = parse_bboxes(
        raw_prediction
    )

    predicted_boxes = [
        normalize_box(box)
        for box in predicted_boxes
    ]

    # Clean textual answer
    prediction = clean_report(
        raw_prediction
    )

    # Text
    bleu1, bleu4 = calculate_bleu(
        reference,
        prediction
    )

    rougeL = calculate_rouge(
        reference,
        prediction
    )

    # Localization
    iou = best_box_iou(
        predicted_boxes,
        gt_boxes
    )

    records.append({

        "index":
            idx,

        "image_path":
            sample["image_path"],

        "question":
            sample["question"],

        "reference":
            reference,

        "prediction":
            prediction,

        "prediction_raw":
            raw_prediction,

        "gt_bboxes":
            json.dumps(gt_boxes),

        "pred_bboxes":
            json.dumps(predicted_boxes),

        "bbox_found":
            int(len(predicted_boxes) > 0),

        "iou":
            float(iou),

        "iou_50":
            int(iou >= 0.50),

        "iou_75":
            int(iou >= 0.75),

        "iou_90":
            int(iou >= 0.90),

        "bleu1":
            float(bleu1),

        "bleu4":
            float(bleu4),

        "rougeL":
            float(rougeL),
    })


results_df = pd.DataFrame(
    records
)

RESULTS_CSV = os.path.join(
    OUTPUT_DIR,
    "qwen3b_test_predictions.csv"
)

results_df.to_csv(
    RESULTS_CSV,
    index=False
)

print(
    f"\nSaved predictions:\n{RESULTS_CSV}"
)


# ------------------------------------------------------------
# 12. BOOTSTRAP 95% CI
# ------------------------------------------------------------

def bootstrap_ci(
    values,
    n_bootstrap=2000
):

    values = np.asarray(
        values,
        dtype=float
    )

    values = values[
        np.isfinite(values)
    ]

    rng = np.random.default_rng(
        SEED
    )

    boot = np.empty(
        n_bootstrap
    )

    for i in range(
        n_bootstrap
    ):

        sample = rng.choice(
            values,
            size=len(values),
            replace=True
        )

        boot[i] = np.mean(
            sample
        )

    return (
        float(np.mean(values)),
        float(np.percentile(boot, 2.5)),
        float(np.percentile(boot, 97.5))
    )


# ------------------------------------------------------------
# 13. BASIC FINAL METRICS
# ------------------------------------------------------------

bleu1_mean, bleu1_low, bleu1_high = (
    bootstrap_ci(
        results_df["bleu1"]
    )
)

bleu4_mean, bleu4_low, bleu4_high = (
    bootstrap_ci(
        results_df["bleu4"]
    )
)

rouge_mean, rouge_low, rouge_high = (
    bootstrap_ci(
        results_df["rougeL"]
    )
)

iou_mean, iou_low, iou_high = (
    bootstrap_ci(
        results_df["iou"]
    )
)

median_iou = float(
    results_df["iou"].median()
)

bbox_rate = float(
    results_df["bbox_found"].mean()
)

iou50_rate = float(
    results_df["iou_50"].mean()
)

iou75_rate = float(
    results_df["iou_75"].mean()
)

iou90_rate = float(
    results_df["iou_90"].mean()
)


# ------------------------------------------------------------
# 14. BERTSCORE
# ------------------------------------------------------------

bert_p = np.nan
bert_r = np.nan
bert_f1 = np.nan

if RUN_BERTSCORE:

    try:

        from bert_score import score

        print("\n" + "=" * 75)
        print("CALCULATING BERTSCORE")
        print("=" * 75)

        references = (
            results_df["reference"]
            .fillna("")
            .astype(str)
            .tolist()
        )

        predictions = (
            results_df["prediction"]
            .fillna("")
            .astype(str)
            .tolist()
        )

        P, R, F1 = score(
            predictions,
            references,
            lang="en",
            model_type="roberta-large",
            device=(
                "cuda"
                if torch.cuda.is_available()
                else "cpu"
            ),
            batch_size=8,
            verbose=True
        )

        bert_p = float(
            P.mean().item()
        )

        bert_r = float(
            R.mean().item()
        )

        bert_f1 = float(
            F1.mean().item()
        )

    except Exception as e:

        print(
            "\nBERTScore skipped:"
        )

        print(e)


# ------------------------------------------------------------
# 15. PRINT FINAL RESULTS
# ------------------------------------------------------------

print("\n")
print("=" * 75)
print("FINAL EVALUATION RESULTS")
print("=" * 75)

print(
    f"Test samples              : {N}"
)

print()

print(
    f"BLEU-1                    : "
    f"{bleu1_mean:.4f}"
)

print(
    f"BLEU-1 95% CI             : "
    f"[{bleu1_low:.4f}, {bleu1_high:.4f}]"
)

print()

print(
    f"BLEU-4                    : "
    f"{bleu4_mean:.4f}"
)

print(
    f"BLEU-4 95% CI             : "
    f"[{bleu4_low:.4f}, {bleu4_high:.4f}]"
)

print()

print(
    f"ROUGE-L                   : "
    f"{rouge_mean:.4f}"
)

print(
    f"ROUGE-L 95% CI            : "
    f"[{rouge_low:.4f}, {rouge_high:.4f}]"
)

print()

print(
    f"BERTScore Precision       : "
    f"{bert_p:.4f}"
)

print(
    f"BERTScore Recall          : "
    f"{bert_r:.4f}"
)

print(
    f"BERTScore F1              : "
    f"{bert_f1:.4f}"
)

print()

print(
    f"Mean IoU                  : "
    f"{iou_mean:.4f}"
)

print(
    f"Mean IoU 95% CI           : "
    f"[{iou_low:.4f}, {iou_high:.4f}]"
)

print(
    f"Median IoU                : "
    f"{median_iou:.4f}"
)

print()

print(
    f"BBox Found                : "
    f"{bbox_rate * 100:.2f}%"
)

print(
    f"IoU >= 0.50               : "
    f"{iou50_rate * 100:.2f}%"
)

print(
    f"IoU >= 0.75               : "
    f"{iou75_rate * 100:.2f}%"
)

print(
    f"IoU >= 0.90               : "
    f"{iou90_rate * 100:.2f}%"
)

print("=" * 75)


# ------------------------------------------------------------
# 16. FINAL PAPER-READY TABLE
# ------------------------------------------------------------

summary = pd.DataFrame({

    "Metric": [

        "BLEU-1",
        "BLEU-4",
        "ROUGE-L",

        "BERTScore Precision",
        "BERTScore Recall",
        "BERTScore F1",

        "Mean IoU",
        "Median IoU",

        "IoU@0.50",
        "IoU@0.75",
        "IoU@0.90",

        "BBox Found (%)",
    ],

    "Qwen2.5-VL-3B": [

        bleu1_mean,
        bleu4_mean,
        rouge_mean,

        bert_p,
        bert_r,
        bert_f1,

        iou_mean,
        median_iou,

        iou50_rate,
        iou75_rate,
        iou90_rate,

        bbox_rate * 100,
    ]
})


print("\n")
print("=" * 75)
print("PAPER-READY SUMMARY")
print("=" * 75)

display(
    summary.style.format(
        {
            "Qwen2.5-VL-3B":
                "{:.4f}"
        }
    )
)


# ------------------------------------------------------------
# 17. SAVE SUMMARY
# ------------------------------------------------------------

SUMMARY_CSV = os.path.join(
    OUTPUT_DIR,
    "qwen3b_final_summary.csv"
)

SUMMARY_JSON = os.path.join(
    OUTPUT_DIR,
    "qwen3b_final_metrics.json"
)

summary.to_csv(
    SUMMARY_CSV,
    index=False
)

metrics_json = {

    "model":
        MODEL_ID,

    "adapter":
        ADAPTER,

    "test_samples":
        N,

    "BLEU-1":
        bleu1_mean,

    "BLEU-1_CI95":
        [
            bleu1_low,
            bleu1_high
        ],

    "BLEU-4":
        bleu4_mean,

    "BLEU-4_CI95":
        [
            bleu4_low,
            bleu4_high
        ],

    "ROUGE-L":
        rouge_mean,

    "ROUGE-L_CI95":
        [
            rouge_low,
            rouge_high
        ],

    "BERTScore_P":
        bert_p,

    "BERTScore_R":
        bert_r,

    "BERTScore_F1":
        bert_f1,

    "Mean_IoU":
        iou_mean,

    "Mean_IoU_CI95":
        [
            iou_low,
            iou_high
        ],

    "Median_IoU":
        median_iou,

    "BBox_Found_Percent":
        bbox_rate * 100,

    "IoU_0.50_Percent":
        iou50_rate * 100,

    "IoU_0.75_Percent":
        iou75_rate * 100,

    "IoU_0.90_Percent":
        iou90_rate * 100,
}

with open(
    SUMMARY_JSON,
    "w"
) as f:

    json.dump(
        metrics_json,
        f,
        indent=4
    )


# ------------------------------------------------------------
# 18. BAD LOCALIZATION CASES
# ------------------------------------------------------------

bad_cases = (
    results_df[
        results_df["iou"] < 0.50
    ]
    .sort_values(
        "iou",
        ascending=True
    )
)

BAD_CASES_CSV = os.path.join(
    OUTPUT_DIR,
    "bad_localization_cases.csv"
)

bad_cases.to_csv(
    BAD_CASES_CSV,
    index=False
)


# ------------------------------------------------------------
# 19. FINAL OUTPUT LOCATIONS
# ------------------------------------------------------------

print("\n")
print("=" * 75)
print("FILES SAVED")
print("=" * 75)

print(
    "1. Per-image results:"
)

print(
    RESULTS_CSV
)

print()

print(
    "2. Final summary:"
)

print(
    SUMMARY_CSV
)

print()

print(
    "3. JSON metrics:"
)

print(
    SUMMARY_JSON
)

print()

print(
    "4. Bad localization cases:"
)

print(
    BAD_CASES_CSV
)

print("=" * 75)
print("EVALUATION COMPLETE")
print("=" * 75)
```

```python
# ============================================================
# ONE-CELL CHEXBERT EVALUATION
# For your already-generated Qwen2.5-VL-3B results
#
# Calculates:
#   CheXbert Precision / Recall / F1 for all 14 findings
#   Macro F1-14
#   Micro F1-14
#   Macro F1-5
#   Micro F1-5
#   Per-finding results
#
# Input:
#   /workspace/insight_cxr_3b/evaluation/qwen3b_test_predictions.csv
#
# Output:
#   CheXbert labeled GT
#   CheXbert labeled predictions
#   Per-finding CSV
#   Final summary CSV
#   Final JSON
# ============================================================

import os
import sys
import subprocess
import importlib.util
import pandas as pd
import numpy as np
import json
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

# ============================================================
# 1. CONFIGURATION
# ============================================================

RESULTS_CSV = (
    "/workspace/insight_cxr_3b/evaluation/"
    "qwen3b_test_predictions.csv"
)

# ------------------------------------------------------------
# IMPORTANT:
# Change this if your CheXbert checkpoint is elsewhere.
#
# Example:
# /workspace/CheXbert/checkpoint/chexbert.pth
# ------------------------------------------------------------

CHEXBERT_REPO = "/workspace/CheXbert"

CHEXBERT_CHECKPOINT = (
    "/workspace/CheXbert/"
    "checkpoint/chexbert.pth"
)

OUTPUT_DIR = (
    "/workspace/insight_cxr_3b/"
    "evaluation/chexbert/"
)

os.makedirs(
    OUTPUT_DIR,
    exist_ok=True
)

# ============================================================
# 2. CHECK INPUT
# ============================================================

print("=" * 75)
print("CHEXBERT EVALUATION")
print("=" * 75)

if not os.path.exists(RESULTS_CSV):

    raise FileNotFoundError(
        "\nPrediction CSV not found:\n"
        + RESULTS_CSV
        + "\n\nRun your Qwen evaluation cell first."
    )

print(
    "Prediction file found:"
)

print(
    RESULTS_CSV
)

# ============================================================
# 3. LOAD YOUR QWEN RESULTS
# ============================================================

df = pd.read_csv(
    RESULTS_CSV
)

print()
print(
    f"Loaded {len(df)} test cases"
)

print()
print(
    "Available columns:"
)

print(
    df.columns.tolist()
)

# Check required columns
required_columns = [
    "reference",
    "prediction"
]

for col in required_columns:

    if col not in df.columns:

        raise ValueError(
            f"Required column '{col}' "
            f"not found in CSV."
        )

# ============================================================
# 4. REMOVE EMPTY REPORTS
# ============================================================

df["reference"] = (
    df["reference"]
    .fillna("")
    .astype(str)
)

df["prediction"] = (
    df["prediction"]
    .fillna("")
    .astype(str)
)

valid_mask = (
    (df["reference"].str.strip() != "")
    &
    (df["prediction"].str.strip() != "")
)

eval_df = df[
    valid_mask
].reset_index(
    drop=True
)

print()
print(
    f"Valid report pairs: "
    f"{len(eval_df)} / {len(df)}"
)

# ============================================================
# 5. CREATE CHEXBERT INPUT FILES
# ============================================================

GT_INPUT = os.path.join(
    OUTPUT_DIR,
    "ground_truth_reports.csv"
)

PRED_INPUT = os.path.join(
    OUTPUT_DIR,
    "generated_reports.csv"
)

gt_df = pd.DataFrame({
    "Report Impression":
        eval_df["reference"]
})

pred_df = pd.DataFrame({
    "Report Impression":
        eval_df["prediction"]
})

gt_df.to_csv(
    GT_INPUT,
    index=False
)

pred_df.to_csv(
    PRED_INPUT,
    index=False
)

print()
print(
    "CheXbert input files created."
)

print(
    GT_INPUT
)

print(
    PRED_INPUT
)

# ============================================================
# 6. CHECK CHEXBERT REPOSITORY
# ============================================================

if not os.path.exists(CHEXBERT_REPO):

    print()
    print(
        "CheXbert repository not found."
    )

    print(
        "Cloning official repository..."
    )

    subprocess.run(
        [
            "git",
            "clone",
            "https://github.com/"
            "stanfordmlgroup/CheXbert.git",
            CHEXBERT_REPO
        ],
        check=True
    )

else:

    print()
    print(
        "CheXbert repository found:"
    )

    print(
        CHEXBERT_REPO
    )

# ============================================================
# 7. LOCATE label.py
# ============================================================

possible_label_files = [

    os.path.join(
        CHEXBERT_REPO,
        "label.py"
    ),

    os.path.join(
        CHEXBERT_REPO,
        "src",
        "label.py"
    )
]

LABEL_SCRIPT = None

for path in possible_label_files:

    if os.path.exists(path):

        LABEL_SCRIPT = path
        break

if LABEL_SCRIPT is None:

    raise FileNotFoundError(
        "\nCould not find CheXbert label.py.\n"
        "Expected either:\n"
        + possible_label_files[0]
        + "\nor\n"
        + possible_label_files[1]
    )

print()
print(
    "CheXbert label script:"
)

print(
    LABEL_SCRIPT
)

# ============================================================
# 8. CHECKPOINT
# ============================================================

if not os.path.exists(
    CHEXBERT_CHECKPOINT
):

    print()
    print("=" * 75)
    print("CHEXBERT CHECKPOINT NOT FOUND")
    print("=" * 75)

    print(
        "\nExpected checkpoint:"
    )

    print(
        CHEXBERT_CHECKPOINT
    )

    print(
        "\nDownload the official CheXbert checkpoint "
        "from the Stanford CheXbert repository and "
        "set CHEXBERT_CHECKPOINT to its .pth path."
    )

    print(
        "\nOfficial repository:"
    )

    print(
        "https://github.com/stanfordmlgroup/CheXbert"
    )

    raise FileNotFoundError(
        "CheXbert checkpoint not found."
    )

print()
print(
    "CheXbert checkpoint found."
)

# ============================================================
# 9. CREATE OUTPUT DIRECTORIES
# ============================================================

GT_OUTPUT = os.path.join(
    OUTPUT_DIR,
    "gt_labeled"
)

PRED_OUTPUT = os.path.join(
    OUTPUT_DIR,
    "pred_labeled"
)

os.makedirs(
    GT_OUTPUT,
    exist_ok=True
)

os.makedirs(
    PRED_OUTPUT,
    exist_ok=True
)

# ============================================================
# 10. RUN CHEXBERT ON GROUND TRUTH
# ============================================================

GT_LABELED = os.path.join(
    GT_OUTPUT,
    "labeled_reports.csv"
)

if not os.path.exists(GT_LABELED):

    print()
    print("=" * 75)
    print("LABELING GROUND-TRUTH REPORTS")
    print("=" * 75)

    command_gt = [
        sys.executable,
        LABEL_SCRIPT,
        f"-d={GT_INPUT}",
        f"-o={GT_OUTPUT}",
        f"-c={CHEXBERT_CHECKPOINT}"
    ]

    print(
        "Running CheXbert..."
    )

    subprocess.run(
        command_gt,
        check=True,
        cwd=os.path.dirname(
            LABEL_SCRIPT
        )
    )

else:

    print()
    print(
        "Ground-truth CheXbert labels already exist."
    )

# ============================================================
# 11. RUN CHEXBERT ON GENERATED REPORTS
# ============================================================

PRED_LABELED = os.path.join(
    PRED_OUTPUT,
    "labeled_reports.csv"
)

if not os.path.exists(PRED_LABELED):

    print()
    print("=" * 75)
    print("LABELING GENERATED REPORTS")
    print("=" * 75)

    command_pred = [
        sys.executable,
        LABEL_SCRIPT,
        f"-d={PRED_INPUT}",
        f"-o={PRED_OUTPUT}",
        f"-c={CHEXBERT_CHECKPOINT}"
    ]

    print(
        "Running CheXbert..."
    )

    subprocess.run(
        command_pred,
        check=True,
        cwd=os.path.dirname(
            LABEL_SCRIPT
        )
    )

else:

    print()
    print(
        "Generated-report CheXbert labels "
        "already exist."
    )

# ============================================================
# 12. LOAD CHEXBERT LABELS
# ============================================================

if not os.path.exists(
    GT_LABELED
):

    raise FileNotFoundError(
        "Ground-truth labeled_reports.csv "
        "was not created."
    )

if not os.path.exists(
    PRED_LABELED
):

    raise FileNotFoundError(
        "Prediction labeled_reports.csv "
        "was not created."
    )

gt_labels = pd.read_csv(
    GT_LABELED
)

pred_labels = pd.read_csv(
    PRED_LABELED
)

print()
print(
    "=" * 75
)

print(
    "CHEXBERT LABELS LOADED"
)

print(
    "=" * 75
)

print(
    "GT shape:",
    gt_labels.shape
)

print(
    "Prediction shape:",
    pred_labels.shape
)

# ============================================================
# 13. CHEXBERT 14 OBSERVATIONS
# ============================================================

OBSERVATIONS = [

    "Fracture",

    "Consolidation",

    "Enlarged Cardiomediastinum",

    "No Finding",

    "Pleural Other",

    "Cardiomegaly",

    "Pneumothorax",

    "Atelectasis",

    "Support Devices",

    "Edema",

    "Pleural Effusion",

    "Lung Lesion",

    "Lung Opacity",
]

# ============================================================
# 14. VERIFY COLUMNS
# ============================================================

print()
print(
    "Checking 14 CheXbert observations..."
)

missing_gt = [
    x for x in OBSERVATIONS
    if x not in gt_labels.columns
]

missing_pred = [
    x for x in OBSERVATIONS
    if x not in pred_labels.columns
]

if missing_gt:

    print(
        "Missing from GT:",
        missing_gt
    )

if missing_pred:

    print(
        "Missing from predictions:",
        missing_pred
    )

available_observations = [
    x for x in OBSERVATIONS
    if (
        x in gt_labels.columns
        and
        x in pred_labels.columns
    )
]

print(
    f"Available observations: "
    f"{len(available_observations)}/14"
)

if len(available_observations) < 14:

    raise ValueError(
        "Not all 14 CheXbert observations "
        "were found."
    )

# ============================================================
# 15. CONVERT CHEXBERT LABELS
#
# Official convention:
#   1    = Positive
#   0    = Negative
#   -1   = Uncertain
#   NaN  = Blank
#
# Standard report-generation F1 evaluation:
#   Positive = 1
#   Everything else = 0
#
# This is the convention commonly used in
# CXR report-generation evaluation.
# ============================================================

def positive_label(series):

    numeric = pd.to_numeric(
        series,
        errors="coerce"
    )

    return (
        numeric == 1
    ).astype(int)


# ============================================================
# 16. PER-FINDING EVALUATION
# ============================================================

rows = []

for finding in available_observations:

    y_true = positive_label(
        gt_labels[finding]
    )

    y_pred = positive_label(
        pred_labels[finding]
    )

    precision = precision_score(
        y_true,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        y_pred,
        zero_division=0
    )

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1]
    ).ravel()

    rows.append({

        "Finding":
            finding,

        "Precision":
            precision,

        "Recall":
            recall,

        "F1":
            f1,

        "TP":
            int(tp),

        "FP":
            int(fp),

        "FN":
            int(fn),

        "TN":
            int(tn),

        "GT_Positive":
            int(y_true.sum()),

        "Pred_Positive":
            int(y_pred.sum()),
    })


per_finding = pd.DataFrame(
    rows
)

# ============================================================
# 17. MACRO F1-14
# ============================================================

macro_f1_14 = float(
    per_finding["F1"].mean()
)

# ============================================================
# 18. MICRO F1-14
# ============================================================

all_true = []
all_pred = []

for finding in available_observations:

    y_true = positive_label(
        gt_labels[finding]
    )

    y_pred = positive_label(
        pred_labels[finding]
    )

    all_true.extend(
        y_true.tolist()
    )

    all_pred.extend(
        y_pred.tolist()
    )


micro_f1_14 = float(
    f1_score(
        all_true,
        all_pred,
        zero_division=0
    )
)

# ============================================================
# 19. FIVE COMMON CXR FINDINGS
#
# Standard F1-5:
#   Atelectasis
#   Cardiomegaly
#   Consolidation
#   Edema
#   Pleural Effusion
# ============================================================

FIVE_FINDINGS = [

    "Atelectasis",

    "Cardiomegaly",

    "Consolidation",

    "Edema",

    "Pleural Effusion",
]

five_rows = []

five_true = []
five_pred = []

for finding in FIVE_FINDINGS:

    y_true = positive_label(
        gt_labels[finding]
    )

    y_pred = positive_label(
        pred_labels[finding]
    )

    f1 = f1_score(
        y_true,
        y_pred,
        zero_division=0
    )

    precision = precision_score(
        y_true,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        zero_division=0
    )

    five_rows.append({

        "Finding":
            finding,

        "Precision":
            precision,

        "Recall":
            recall,

        "F1":
            f1,
    })

    five_true.extend(
        y_true.tolist()
    )

    five_pred.extend(
        y_pred.tolist()
    )


five_results = pd.DataFrame(
    five_rows
)

macro_f1_5 = float(
    five_results["F1"].mean()
)

micro_f1_5 = float(
    f1_score(
        five_true,
        five_pred,
        zero_division=0
    )
)

# ============================================================
# 20. CHEXBERT AVERAGE
# ============================================================

chexbert_average = float(
    np.mean([
        macro_f1_14,
        micro_f1_14,
        macro_f1_5,
        micro_f1_5
    ])
)

# ============================================================
# 21. PRINT 14-FINDING RESULTS
# ============================================================

print()
print("=" * 100)
print("CHEXBERT — PER-FINDING RESULTS")
print("=" * 100)

display(
    per_finding.style.format(
        {
            "Precision": "{:.4f}",
            "Recall": "{:.4f}",
            "F1": "{:.4f}",
        }
    )
)

# ============================================================
# 22. PRINT FIVE-FINDING RESULTS
# ============================================================

print()
print("=" * 100)
print("CHEXBERT — FIVE COMMON FINDINGS")
print("=" * 100)

display(
    five_results.style.format(
        {
            "Precision": "{:.4f}",
            "Recall": "{:.4f}",
            "F1": "{:.4f}",
        }
    )
)

# ============================================================
# 23. FINAL CHEXBERT RESULTS
# ============================================================

print()
print("=" * 100)
print("FINAL CHEXBERT RESULTS")
print("=" * 100)

print(
    f"Test reports              : "
    f"{len(eval_df)}"
)

print()

print(
    f"Macro F1 (14)              : "
    f"{macro_f1_14:.4f}"
)

print(
    f"Micro F1 (14)              : "
    f"{micro_f1_14:.4f}"
)

print()

print(
    f"Macro F1 (5)               : "
    f"{macro_f1_5:.4f}"
)

print(
    f"Micro F1 (5)               : "
    f"{micro_f1_5:.4f}"
)

print()

print(
    f"Overall CheXbert Average   : "
    f"{chexbert_average:.4f}"
)

print("=" * 100)

# ============================================================
# 24. SAVE PER-FINDING RESULTS
# ============================================================

PER_FINDING_CSV = os.path.join(
    OUTPUT_DIR,
    "chexbert_per_finding.csv"
)

FIVE_CSV = os.path.join(
    OUTPUT_DIR,
    "chexbert_five_findings.csv"
)

per_finding.to_csv(
    PER_FINDING_CSV,
    index=False
)

five_results.to_csv(
    FIVE_CSV,
    index=False
)

# ============================================================
# 25. CREATE PAPER-READY SUMMARY
# ============================================================

paper_summary = pd.DataFrame({

    "Metric": [

        "CheXbert Macro F1 (14)",

        "CheXbert Micro F1 (14)",

        "CheXbert Macro F1 (5)",

        "CheXbert Micro F1 (5)",

        "CheXbert Average",
    ],

    "Qwen2.5-VL-3B": [

        macro_f1_14,

        micro_f1_14,

        macro_f1_5,

        micro_f1_5,

        chexbert_average,
    ]
})

print()
print("=" * 100)
print("PAPER-READY CHEXBERT TABLE")
print("=" * 100)

display(
    paper_summary.style.format(
        {
            "Qwen2.5-VL-3B":
                "{:.4f}"
        }
    )
)

# ============================================================
# 26. SAVE SUMMARY
# ============================================================

SUMMARY_CSV = os.path.join(
    OUTPUT_DIR,
    "chexbert_final_summary.csv"
)

SUMMARY_JSON = os.path.join(
    OUTPUT_DIR,
    "chexbert_final_metrics.json"
)

paper_summary.to_csv(
    SUMMARY_CSV,
    index=False
)

final_json = {

    "test_samples":
        int(len(eval_df)),

    "chexbert_macro_f1_14":
        macro_f1_14,

    "chexbert_micro_f1_14":
        micro_f1_14,

    "chexbert_macro_f1_5":
        macro_f1_5,

    "chexbert_micro_f1_5":
        micro_f1_5,

    "chexbert_average":
        chexbert_average,

    "chexbert_observations":
        available_observations,
}

with open(
    SUMMARY_JSON,
    "w"
) as f:

    json.dump(
        final_json,
        f,
        indent=4
    )

# ============================================================
# 27. FINAL FILE LOCATIONS
# ============================================================

print()
print("=" * 100)
print("CHEXBERT EVALUATION COMPLETE")
print("=" * 100)

print(
    "\n1. Ground-truth labels:"
)

print(
    GT_LABELED
)

print(
    "\n2. Generated-report labels:"
)

print(
    PRED_LABELED
)

print(
    "\n3. Per-finding results:"
)

print(
    PER_FINDING_CSV
)

print(
    "\n4. Five-finding results:"
)

print(
    FIVE_CSV
)

print(
    "\n5. Paper summary:"
)

print(
    SUMMARY_CSV
)

print(
    "\n6. JSON:"
)

print(
    SUMMARY_JSON
)

print("=" * 100)
```

```python
import os, json, re, torch
from PIL import Image
from tqdm import tqdm
from peft import PeftModel
from transformers import Qwen2_5_VLForConditionalGeneration, AutoProcessor, BitsAndBytesConfig
import pandas as pd
import numpy as np
from nltk.translate.bleu_score import corpus_bleu, SmoothingFunction
from rouge_score import rouge_scorer

os.environ["TRANSFORMERS_NO_FLASH_ATTN"] = "1"

ADAPTER_DIR = "/workspace/insight_cxr_3b/adapter/"
MODEL_ID    = "Qwen/Qwen2.5-VL-3B-Instruct"
TEST_JSONL  = "/workspace/insight_cxr/processed/test.jsonl"
OUTPUT_DIR  = "/workspace/insight_cxr_3b/evaluation/full_test/"
os.makedirs(OUTPUT_DIR, exist_ok=True)

SYSTEM_PROMPT = (
    "You are an expert radiologist AI assistant. "
    "Given a chest X-ray image, provide accurate clinical findings. "
    "When referring to a specific region, output its bounding box "
    "immediately after using the format: <box>[x1, y1, x2, y2]</box> "
    "where coordinates are absolute pixels on a 448x448 image. "
    "Be factual. Do not hallucinate findings not visible in the image."
)

# ── Load model ────────────────────────────────────────────────
print("Loading 3B model...")
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_compute_dtype=torch.bfloat16,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_use_double_quant=True,
)
base_model = Qwen2_5_VLForConditionalGeneration.from_pretrained(
    MODEL_ID,
    quantization_config=bnb_config,
    device_map="auto",
    torch_dtype=torch.bfloat16,
    attn_implementation="eager",
)
base_model.config.use_cache = True
ft_model  = PeftModel.from_pretrained(base_model, ADAPTER_DIR)
ft_model.eval()
processor = AutoProcessor.from_pretrained(
    ADAPTER_DIR,
    min_pixels=256*28*28,
    max_pixels=1280*28*28,
    padding_side="left",
)
if processor.tokenizer.pad_token is None:
    processor.tokenizer.pad_token = processor.tokenizer.eos_token
print("Model ready.")

# ── Load all test records ─────────────────────────────────────
with open(TEST_JSONL) as f:
    all_test = [json.loads(l) for l in f if l.strip()]
print(f"Total test records: {len(all_test):,}")

# ── Helper functions ──────────────────────────────────────────
def compute_iou(b1, b2):
    x1 = max(b1[0], b2[0]); y1 = max(b1[1], b2[1])
    x2 = min(b1[2], b2[2]); y2 = min(b1[3], b2[3])
    inter = max(0, x2-x1) * max(0, y2-y1)
    a1 = (b1[2]-b1[0]) * (b1[3]-b1[1])
    a2 = (b2[2]-b2[0]) * (b2[3]-b2[1])
    union = a1 + a2 - inter
    return inter / union if union > 0 else 0.0

def run_inference(image_path, question):
    try:
        img = Image.open(image_path).convert("RGB")
    except:
        return ""
    messages = [
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user", "content": [
            {"type": "image"},
            {"type": "text", "text": question},
        ]},
    ]
    prompt = processor.apply_chat_template(
        messages, tokenize=False, add_generation_prompt=True
    )
    inputs = processor(
        text=[prompt], images=[img],
        return_tensors="pt", padding=True
    ).to("cuda")
    with torch.no_grad():
        out = ft_model.generate(
            **inputs, max_new_tokens=300,
            do_sample=False, repetition_penalty=1.1,
        )
    return processor.tokenizer.decode(
        out[0][inputs["input_ids"].shape[1]:],
        skip_special_tokens=True,
    )

# ── Evaluation loop ───────────────────────────────────────────
print(f"\nEvaluating all {len(all_test):,} test records...")
print("This will take several hours. Progress saved every 500 samples.\n")

records     = []
predictions = []
references  = []
iou_scores  = []
bbox_found  = 0

SAVE_EVERY = 500

for i, sample in enumerate(tqdm(all_test)):
    response = run_inference(sample["image_path"], sample["question"])
    if not response:
        continue

    clean      = re.sub(r'<think>.*?</think>', '', response, flags=re.DOTALL).strip()
    answer     = re.sub(r'<box>\[.*?\]</box>', '', clean).strip()
    pred_boxes = re.findall(
        r'<box>\[(\d+),\s*(\d+),\s*(\d+),\s*(\d+)\]</box>', response
    )

    predictions.append(answer)
    references.append(sample["answer"])

    iou = 0.0
    if pred_boxes and sample["bboxes"]:
        bbox_found += 1
        pred = [int(x) for x in pred_boxes[0]]
        iou  = compute_iou(pred, sample["bboxes"][0])
    elif sample["bboxes"]:
        pass
    iou_scores.append(iou)

    records.append({
        "index"      : i,
        "image_path" : sample["image_path"],
        "question"   : sample["question"],
        "reference"  : sample["answer"],
        "prediction" : answer,
        "gt_bboxes"  : json.dumps(sample["bboxes"]),
        "pred_bboxes": json.dumps(pred_boxes),
        "bbox_found" : int(len(pred_boxes) > 0),
        "iou"        : float(iou),
        "iou_50"     : int(iou >= 0.50),
        "iou_75"     : int(iou >= 0.75),
        "iou_90"     : int(iou >= 0.90),
    })

    # Save intermediate results every 500 samples
    if (i+1) % SAVE_EVERY == 0:
        df_temp = pd.DataFrame(records)
        df_temp.to_csv(OUTPUT_DIR + f"results_checkpoint_{i+1}.csv", index=False)
        curr_iou = sum(iou_scores)/len(iou_scores)
        print(f"\n[{i+1}/{len(all_test)}] mean IoU: {curr_iou:.3f} | bbox_found: {bbox_found}/{i+1}")

# ── Final metrics ─────────────────────────────────────────────
smooth    = SmoothingFunction().method1
tok_preds = [p.lower().split() for p in predictions]
tok_refs  = [[r.lower().split()] for r in references]

bleu1 = corpus_bleu(tok_refs, tok_preds,
                    weights=(1,0,0,0), smoothing_function=smooth)
bleu4 = corpus_bleu(tok_refs, tok_preds,
                    weights=(.25,.25,.25,.25), smoothing_function=smooth)

scorer  = rouge_scorer.RougeScorer(["rougeL"], use_stemmer=True)
rouge_l = sum(scorer.score(r, p)["rougeL"].fmeasure
              for r, p in zip(references, predictions)) / len(predictions)

mean_iou    = sum(iou_scores) / len(iou_scores)
median_iou  = float(np.median(iou_scores))
iou50_rate  = sum(1 for x in iou_scores if x >= 0.50) / len(iou_scores)
iou75_rate  = sum(1 for x in iou_scores if x >= 0.75) / len(iou_scores)
iou90_rate  = sum(1 for x in iou_scores if x >= 0.90) / len(iou_scores)
bbox_rate   = bbox_found / len(iou_scores)

print(f"\n{'='*60}")
print(f"FULL TEST SET EVALUATION  (n={len(all_test):,})")
print(f"{'='*60}")
print(f"BLEU-1        : {bleu1:.4f}")
print(f"BLEU-4        : {bleu4:.4f}")
print(f"ROUGE-L       : {rouge_l:.4f}")
print(f"Mean IoU      : {mean_iou:.4f}")
print(f"Median IoU    : {median_iou:.4f}")
print(f"IoU >= 0.50   : {iou50_rate*100:.2f}%")
print(f"IoU >= 0.75   : {iou75_rate*100:.2f}%")
print(f"IoU >= 0.90   : {iou90_rate*100:.2f}%")
print(f"BBox Found    : {bbox_rate*100:.2f}%")
print(f"{'='*60}")

# ── Save final results ────────────────────────────────────────
df_final = pd.DataFrame(records)
df_final.to_csv(OUTPUT_DIR + "full_test_results.csv", index=False)

metrics = {
    "total_samples" : len(all_test),
    "BLEU-1"        : round(bleu1, 4),
    "BLEU-4"        : round(bleu4, 4),
    "ROUGE-L"       : round(rouge_l, 4),
    "Mean_IoU"      : round(mean_iou, 4),
    "Median_IoU"    : round(median_iou, 4),
    "IoU_0.50"      : round(iou50_rate, 4),
    "IoU_0.75"      : round(iou75_rate, 4),
    "IoU_0.90"      : round(iou90_rate, 4),
    "BBox_Found_Pct": round(bbox_rate*100, 2),
}
with open(OUTPUT_DIR + "full_test_metrics.json", "w") as f:
    json.dump(metrics, f, indent=4)

print(f"\nResults saved → {OUTPUT_DIR}")
print("full_test_results.csv")
print("full_test_metrics.json")
```
